```python
# Code Example 3.22 — A function that flags product line performance
def flag_performance(product, revenue, target):
    if revenue >= target:
        status = 'On target'
    elif revenue >= target * 0.90:
        status = 'Near target (within 10%)'
    else:
        status = 'Below target'
    shortfall = max(0, target - revenue)
    return status, shortfall

product_lines = ['Beverages', 'Snacks', 'Personal Care', 'Household', 'Vitamins']
revenues      = [84750.00, 61200.00, 43800.00, 29400.00, 52100.00]
target        = 60000.00

for product, revenue in zip(product_lines, revenues):
    status, shortfall = flag_performance(product, revenue, target)
    if shortfall > 0:
        print(f"  {product:<15} {status}. Shortfall: ${shortfall:,.2f}")
    else:
        print(f"  {product:<15} {status}.")
```

```python
# Code Example 3.24 — A function with a docstring
def flag_performance(product, revenue, target=60000.00):
    """
    Evaluate a product line's weekly revenue against its target.

    Parameters
    ----------
    product : str
        Name of the product line.
    revenue : float
        Actual revenue for the week.
    target : float, optional
        Weekly revenue target. Default is 60000.00.

    Returns
    -------
    status : str
        Performance classification: 'On target', 'Near target', or
        'Below target'.
    shortfall : float
        Amount below target. Zero if on or above target.
    """
    if revenue >= target:
        status = 'On target'
    elif revenue >= target * 0.90:
        status = 'Near target (within 10%)'
    else:
        status = 'Below target'
    shortfall = max(0, target - revenue)
    return status, shortfall

help(flag_performance)
```

## Section 3.7 — Putting It Together: Maya's Monday Report

Everything above, assembled into three reusable functions. Run the cell
below once to define them — then every Monday reduces to a single call.

```python
# Code Example 3.30 — Three functions that power Maya's Monday report

def clean_revenue(series):
    """Convert a revenue column to float. Missing values become NaN."""
    return (
        series.astype(str)
        .str.replace('[$,]', '', regex=True)
        .replace({'None': np.nan, 'nan': np.nan})
        .astype(float)
    )

def flag_performance(revenue, target):
    """Classify revenue against target. Returns status and shortfall."""
    if pd.isna(revenue):
        return 'Missing data', np.nan
    elif revenue >= target:
        return 'On target', 0.0
    elif revenue >= target * 0.90:
        return 'Near target', round(target - revenue, 2)
    else:
        return 'Below target', round(target - revenue, 2)

def weekly_report(regional_data, target=60000.00, week_label='This week'):
    """
    Produce Maya's Monday sales summary from regional DataFrames.

    Parameters
    ----------
    regional_data : list of tuple
        Each tuple is (DataFrame, region_name).
    target : float, optional
        Weekly revenue target per product line. Default is 60000.00.
    week_label : str, optional
        Label for the report header.

    Returns
    -------
    pd.DataFrame
        Summary table with status and shortfall columns added.
    """
    cleaned = []
    for df, region in regional_data:
        df = df.copy()
        df['revenue'] = clean_revenue(df['revenue'])
        df['units']   = (pd.to_numeric(df['units'], errors='coerce')
                           .fillna(0).astype(int))
        df['region']  = region
        cleaned.append(df)

    weekly = pd.concat(cleaned, ignore_index=True)

    flags = weekly.apply(
        lambda row: flag_performance(row['revenue'], target),
        axis=1, result_type='expand'
    )
    flags.columns = ['status', 'shortfall']
    weekly = pd.concat([weekly, flags], axis=1)

    total_revenue   = weekly['revenue'].sum()
    total_shortfall = weekly['shortfall'].sum()
    n_below  = (weekly['status'] == 'Below target').sum()
    n_lines  = len(weekly)

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    pivot = weekly.pivot(index='product', columns='region', values='revenue')
    pivot.plot(kind='bar', ax=axes[0], colormap='Blues',
               edgecolor='white', alpha=0.85)
    axes[0].axhline(target, color='gray', linestyle='--',
                    linewidth=1, label='Target')
    axes[0].set_title('Revenue by Product Line and Region', fontsize=11)
    axes[0].set_xlabel('')
    axes[0].set_ylabel('Revenue (USD)')
    axes[0].tick_params(axis='x', rotation=30)
    axes[0].legend(fontsize=8)
    shortfall_by_product = weekly.groupby('product')['shortfall'].sum()
    bar_colors = ['tomato' if v > 0 else 'steelblue'
                  for v in shortfall_by_product]
    axes[1].bar(shortfall_by_product.index, shortfall_by_product.values,
                color=bar_colors, edgecolor='white', alpha=0.85)
    axes[1].set_title('Total Shortfall by Product Line', fontsize=11)
    axes[1].set_ylabel('Shortfall (USD)')
    axes[1].tick_params(axis='x', rotation=30)
    plt.suptitle(
        f'Veritas Consumer Goods — Weekly Sales Summary ({week_label})',
        fontsize=13, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

    print("=" * 54)
    print(f"  VERITAS CONSUMER GOODS — {week_label.upper()}")
    print("=" * 54)
    print(f"  Total revenue:      ${total_revenue:>12,.2f}")
    print(f"  Total shortfall:    ${total_shortfall:>12,.2f}")
    print(f"  Lines below target: {n_below} of {n_lines}")
    print("-" * 54)
    print(f"  {'Product':<15} {'Region':<12} {'Revenue':>10}  Status")
    print("-" * 54)
    for _, row in weekly.sort_values(['product', 'region']).iterrows():
        flag = ' ***' if row['status'] == 'Below target' else ''
        rev  = (f"${row['revenue']:>9,.0f}" if not pd.isna(row['revenue'])
                else '   MISSING')
        print(f"  {row['product']:<15} {row['region']:<12}"
              f" {rev}  {row['status']}{flag}")
    print("=" * 54)
    return weekly

print("Functions defined: clean_revenue, flag_performance, weekly_report")
```

```python
# Code Example 3.31 — Running the weekly report: the only call Maya needs each Monday

# Monday morning: load the three files and call the function.
# Everything else — cleaning, merging, flagging, charting, reporting —
# is handled automatically inside weekly_report().
northeast = pd.DataFrame({
    'product': ['Beverages','Snacks','Personal Care','Household','Vitamins'],
    'revenue': ['$84,750.00','$61,200.00','$43,800.00',
                '$29,400.00','$52,100.00'],
    'units':   [1420, 980, 730, 420, 890],
})
midwest = pd.DataFrame({
    'product': ['Beverages','Snacks','Personal Care','Household','Vitamins'],
    'revenue': ['72300','55800','38400', None,'47600'],
    'units':   [1210, 890, 640, None, 760],
})
south = pd.DataFrame({
    'product': ['Beverages','Snacks','Personal Care','Household','Vitamins'],
    'revenue': ['68500.00','49200.00','41100.00','33800.00','44900.00'],
    'units':   [1150, 830, 685, 560, 720],
})

# Three lines. One function call. The full report.
summary = weekly_report(
    regional_data=[(northeast, 'Northeast'),
                   (midwest,   'Midwest'),
                   (south,     'South')],
    target=60000.00,
    week_label='Week 42'
)
```

Try changing the `target`, adding a fourth region, or relabeling the week —
the function adapts without any changes to its internal logic. That is the
practical payoff of writing `weekly_report()` as a function instead of a
one-off script.
